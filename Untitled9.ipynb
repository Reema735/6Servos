```python

import pandas as pd
import matplotlib.pyplot as plt

PROJECT_NAME = "IT Service Desk Ticket Prediction"

df = pd.DataFrame({
    "Priority": [
        1,2,1,3,2,1,2,1,3,2,1,2,
        1,2,3,1,2,3,1,2,1,3,2,1
    ],
    "Open_Hours": [
        2,18,5,36,12,3,25,7,44,10,4,15,
        6,20,50,8,22,60,3,30,9,48,28,5
    ],
    "Attempts": [
        1,2,1,4,2,1,3,1,5,2,1,2,
        1,3,5,1,3,5,1,4,2,4,3,1
    ],
    "Category": [
        "Software","Software","Software","Network",
        "Software","Software","Network","Software",
        "Hardware","Software","Software","Network",
        "Software","Network","Hardware","Software",
        "Network","Hardware","Software","Network",
        "Software","Hardware","Network","Network"
    ],
    "Resolved": [
        1,1,1,0,1,1,0,1,0,1,1,1,
        1,0,0,1,0,0,1,0,1,0,0,1
    ]
})

print("Project:", PROJECT_NAME)
print("Total Tickets:", len(df))
print("\nTicket Status:\n", df["Resolved"].value_counts())

# Visualizations
fig, axes = plt.subplots(1, 3, figsize=(15, 4))

status = df["Resolved"].value_counts().reindex(
    [1, 0], fill_value=0
)

axes[0].pie(
    status,
    labels=["Resolved", "Unresolved"],
    autopct="%1.1f%%",
    colors=["#2ecc71", "#e74c3c"]
)
axes[0].set_title("Ticket Resolution Status")

categories = df["Category"].value_counts().reindex(
    ["Software", "Network", "Hardware"]
)

axes[1].bar(
    categories.index,
    categories.values,
    color=["#3498db", "#9b59b6", "#f39c12"]
)
axes[1].set_title("Tickets by Category")
axes[1].set_ylabel("Number of Tickets")

priorities = df["Priority"].value_counts().reindex([1,2,3])

axes[2].bar(
    ["Low", "Medium", "High"],
    priorities.values,
    color=["#2ecc71", "#f1c40f", "#e74c3c"]
)
axes[2].set_title("Tickets by Priority")
axes[2].set_ylabel("Number of Tickets")

plt.tight_layout()
plt.show()

print("✅ Code 1 Completed!")
```

```python

from sklearn.model_selection import train_test_split

# Feature Engineering
df["Hours_Per_Priority"] = (
    df["Open_Hours"] / (df["Priority"] + 1)
)

# Features and Target
X = df.drop(columns=["Resolved"])
y = df["Resolved"]

# Train / Test Split
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.25,
    random_state=42,
    stratify=y
)

print("Training Records:", len(X_train))
print("Testing Records:", len(X_test))
print("\nNew Features:\n", X_train.head())
print("✅ Code 2 Completed!")
```

```python

from sklearn.pipeline import make_pipeline
from sklearn.compose import make_column_transformer
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier

# Preprocessing
preprocess = make_column_transformer(
    (
        make_pipeline(
            SimpleImputer(strategy="median"),
            StandardScaler()
        ),
        ["Priority", "Open_Hours", "Attempts",
         "Hours_Per_Priority"]
    ),
    (
        make_pipeline(
            SimpleImputer(strategy="most_frequent"),
            OneHotEncoder(handle_unknown="ignore")
        ),
        ["Category"]
    )
)

# Models
models = {
    "Logistic Regression": make_pipeline(
        preprocess, LogisticRegression(max_iter=1000)
    ),
    "Random Forest": make_pipeline(
        preprocess, RandomForestClassifier(random_state=42)
    ),
    "Gradient Boosting": make_pipeline(
        preprocess, GradientBoostingClassifier(random_state=42)
    )
}

for name, model in models.items():
    model.fit(X_train, y_train)
    print("✅ Trained:", name)

print("✅ Code 3 Completed!")
```

```python

from sklearn.model_selection import GridSearchCV

params = {
    "gradientboostingclassifier__n_estimators": [20, 50],
    "gradientboostingclassifier__learning_rate": [0.05, 0.1],
    "gradientboostingclassifier__max_depth": [1, 2]
}

search = GridSearchCV(
    models["Gradient Boosting"],
    params,
    scoring="average_precision",
    cv=3
)

search.fit(X_train, y_train)

best_model = search.best_estimator_
models["Gradient Boosting"] = best_model

print("Best Parameters:", search.best_params_)
print("Best CV Score:", round(search.best_score_, 2))
print("✅ Code 4 Completed!")
```

```python

from sklearn.metrics import (
    accuracy_score,
    f1_score,
    average_precision_score
)

# Evaluate Models
results = []

for name, model in models.items():

    predictions = model.predict(X_test)
    probabilities = model.predict_proba(X_test)[:, 1]

    results.append({
        "Model": name,
        "Accuracy": accuracy_score(y_test, predictions),
        "F1 Score": f1_score(
            y_test, predictions, zero_division=0
        ),
        "Average Precision": average_precision_score(
            y_test, probabilities
        )
    })

# Model Comparison Table
results_df = pd.DataFrame(results)

print("MODEL COMPARISON")
print(results_df.round(2))

# Model Comparison Chart
results_df.set_index("Model").plot(
    kind="bar",
    figsize=(10, 5),
    color=["#3498db", "#2ecc71", "#f39c12"],
    rot=0
)

plt.title("Machine Learning Model Comparison")
plt.ylabel("Score")
plt.xlabel("Models")
plt.ylim(0, 1.05)
plt.legend()
plt.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.show()

print("✅ Code 5 Completed!")
```

```python

# Feature Importance
feature_names = best_model[0].get_feature_names_out()

importance = pd.Series(
    best_model[-1].feature_importances_,
    index=feature_names
).sort_values()

plt.figure(figsize=(9, 5))
importance.plot(kind="barh", color="#3498db")
plt.title("Feature Importance")
plt.xlabel("Importance")
plt.tight_layout()
plt.show()

print("Most Important Feature:", importance.idxmax())

# New IT Ticket
new_ticket = pd.DataFrame({
    "Priority": [2],
    "Open_Hours": [5],
    "Attempts": [1],
    "Category": ["Software"]
})

new_ticket["Hours_Per_Priority"] = (
    new_ticket["Open_Hours"] /
    (new_ticket["Priority"] + 1)
)

# Prediction
probability = best_model.predict_proba(new_ticket)[0][1]
prediction = int(probability >= 0.5)

print("\nNEW TICKET PREDICTION")
print("Model: Gradient Boosting")
print("Prediction:", "Resolved" if prediction == 1 else "Unresolved")
print("Resolution Probability:", round(probability * 100, 2), "%")

# Prediction Chart
plt.figure(figsize=(7, 3))
plt.barh(
    ["New Ticket"],
    [probability * 100],
    color="#2ecc71" if prediction == 1 else "#e74c3c"
)

plt.axvline(50, color="gray", linestyle="--")
plt.xlim(0, 100)
plt.xlabel("Resolution Probability (%)")
plt.title("New Ticket Prediction")
plt.tight_layout()
plt.show()

print("✅ Project Completed!")
```
